```python
def calculate_rental(car_type, days, driver=False):
    if car_type.strip().lower() == "economy":
        rate = 3000
    elif car_type.strip().lower() == "sedan":
        rate = 5000
    elif car_type.strip().lower() == "suv":
        rate = 8000
    else:
        print(f"Error: Unknown car type '{car_type}'")
        return

    car_cost = rate * days

    if driver:
        driver_cost = 2000 * days
    else:
        driver_cost = 0

    subtotal = car_cost + driver_cost

    if days >= 7:
        discount_rate = 0.15
    elif days >= 3:
        discount_rate = 0.10
    else:
        discount_rate = 0.00

    discount = subtotal * discount_rate
    final_bill = subtotal - discount

    print(f"Car Type: {car_type}")
    print(f"Rate per Day: {rate:.0f}")
    print(f"Days: {days}")
    print(f"Car Cost: {car_cost:.0f}")
    print(f"Driver: {'Yes' if driver else 'No'}")
    print(f"Driver Cost: {driver_cost:.0f}")
    print(f"Subtotal: {subtotal:.0f}")
    print(f"Discount: {discount:.0f} ({discount_rate * 100:.0f}%)")
    print(f"Final Bill: {final_bill:.0f}")

car_type = input("Enter car type (Economy/Sedan/SUV): ")
days     = int(input("Enter number of days: "))
drv_in   = input("Driver required? (yes/no): ").strip().lower()
driver   = drv_in == "yes"

calculate_rental(car_type, days, driver)
```
